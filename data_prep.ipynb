Data Preparation

```python
import pandas as pd
import numpy as np 
import pathlib as path 
```

```python
raw = path.Path('./data/raw')
processed = path.Path('./data/processed')
raw.mkdir(parents=True, exist_ok=True)
processed.mkdir(parents=True, exist_ok=True)
```

```python
orders = pd.read_excel(raw/'List_of_orders_1.xlsx',parse_dates=['Order Date'])
details = pd.read_excel(raw/'Order_Details_1.xlsx')
targets = pd.read_excel(raw/'Sales_target_1.xlsx',parse_dates=['Month of Order Date'])
```

```python
def clean_cols(df):
    df.columns = (df.columns
                  .str.strip()
                  .str.lower()
                  .str.replace(" ","_")
                  .str.replace("\n","_")
                  )
    return df
orders = clean_cols(orders)
details = clean_cols(details)
tergets = clean_cols(targets)

orders.head()
details.head()
targets.head()
```

```python
print("orders info")
orders.info()
print("\ndetails info")
details.info()
print("\mtargets info")
targets.info()
print("\n orders missing vales \n",orders.isna().sum())
print("\ndetails missing values \n",details.isna().sum())
```

```python
merged = pd.merge(orders,details,on='order_id',how='inner',suffixes=('_order','_detail'))
print("merged shape:",merged.shape)
merged.head()
```

```python
numeric_col = ['amount','price','quantity']

for col in numeric_col:
    if col in merged.columns:
        merged[col] = pd.to_numeric(merged[col],errors='coerce').fillna(0)
merged.info()
```

```python
print('Missing values\n:',merged.isna().sum())
print("\nunique Id's:",merged['order_id'].unique())
print("\n range of date:",merged['order_date'].min(),"to",merged['order_date'].max())
```

```python
merged.to_csv(processed/'merged_orders_details.csv',index=False)
```

```python
summary = merged.describe(include="all")
summary.to_csv(processed/'merged_summary.csv',index=False)
```
